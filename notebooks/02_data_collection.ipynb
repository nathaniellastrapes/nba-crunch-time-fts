```python
from nba_api.stats.endpoints import playbyplayv3, leaguegamefinder
from nba_api.stats.static import teams
import pandas as pd
import numpy as np
from nba_crunch.data_processing import enrich_free_throws, parse_clock, get_season_game_ids, pull_and_save_pbp
from pathlib import Path
import time
from tqdm import tqdm

pd.set_option('display.max_columns', None)
```

```python
seasons = ['2023-24', '2024-25', '2025-26']
all_game_ids = []

for season in seasons:
    ids = get_season_game_ids(season)
    print(f"{season}: {len(ids)})")
    all_game_ids.extend(ids)

print(f"Total: {len(all_game_ids)}")

pd.DataFrame({'game_id': all_game_ids}).to_csv('../data/raw/game_ids.csv', index=False)
```

```python
from nba_crunch.data_processing import pull_and_save_pbp

output_dir = Path('../data/raw/pbp')
output_dir.mkdir(parents=True, exist_ok=True)

results = {'pulled': 0, 'skipped': 0, 'failed': []}

for game_id in tqdm(all_game_ids):
    status = pull_and_save_pbp(game_id, output_dir)
    if status == 'failed':
        results['failed'].append(game_id)
    else:
        results[status] += 1

print(results)
```
